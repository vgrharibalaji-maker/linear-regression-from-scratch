```python
import numpy as np

# Step 1: Create the data
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
scores = np.array([35, 40, 50, 55, 65, 70, 78, 85])

print(hours)
print(scores)

# Step 2: Create a prediction function
def predict(hours, m, c):
    return m * hours + c

# Try a guess
test_prediction = predict(hours, 5, 10)
print(test_prediction)
```

```python
def mean_squared_error(actual, predicted):
    errors = actual - predicted
    squared_errors = errors ** 2
    return np.mean(squared_errors)

loss = mean_squared_error(scores, test_prediction)
print(loss)
```

```python
# Step 4: Gradient Descent
m = 0
c = 0
learning_rate = 0.01
epochs = 1000  # number of times we repeat the adjustment

n = len(hours)

for i in range(epochs):
    predicted = predict(hours, m, c)
    error = scores - predicted

    m_gradient = -2 * np.mean(hours * error)
    c_gradient = -2 * np.mean(error)

    m = m - learning_rate * m_gradient
    c = c - learning_rate * c_gradient

    if i % 100 == 0:
        loss = mean_squared_error(scores, predicted)
        print(f"Epoch {i}: loss = {loss:.2f}, m = {m:.2f}, c = {c:.2f}")

print("\nFinal values:")
print("m =", m)
print("c =", c)
```

```python
new_hours = 9
predicted_score = predict(new_hours, m, c)
print(f"Predicted score for {new_hours} hours studied: {predicted_score:.2f}")
```

```python
import matplotlib.pyplot as plt

# Plot actual data points
plt.scatter(hours, scores, color='blue', label='Actual data')

# Plot the learned line
predicted_line = predict(hours, m, c)
plt.plot(hours, predicted_line, color='red', label='Learned line')

plt.xlabel('Hours studied')
plt.ylabel('Exam score')
plt.title('Linear Regression: Hours vs Score')
plt.legend()
plt.show()
```
